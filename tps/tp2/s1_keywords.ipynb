# Extraction de Keywords depuis Le Soir / 1960

## Imports

```python
import os
import yake
```

## Extraire les mots clés d'un document avec Yake

https://github.com/LIAAD/yake

```python
# Instancier l'extracteur de mots-clés
kw_extractor = yake.KeywordExtractor(lan="fr", top=50)
kw_extractor
```

```python
# Année étudiée : 1960 (année indépendance Congo Belge)
annee = "1960"

# Chemin vers les fichiers texte
data_path = "../../data/txt/"

# Sélectionner tous les fichiers de 1960
files = sorted([
    f for f in os.listdir(data_path)
    if f.endswith(".txt") and f"_{annee}-" in f
])

# Afficher le nombre de fichiers trouvés
print("Nombre de fichiers pour 1960 :", len(files))
```

```python
# Imprimer le nombre de fichiers identifiés
len(files)
```

```python
# Afficher tous les fichiers sélectionnés pour 1960
sorted(files)
```

```python
# Sélectionner 15 fichiers répartis sur l'année 1960
nombre_fichiers = 15

indices = [
    round(i * (len(files) - 1) / (nombre_fichiers - 1))
    for i in range(nombre_fichiers)
]

fichiers_analyse = [files[i] for i in indices]

# Afficher les fichiers sélectionnés
fichiers_analyse
```

```python
# Choisir le premier fichier parmi les 15 sélectionnés
this_file = fichiers_analyse[0]
this_file
```

```python
# Récupérer le texte du fichier
text = open(os.path.join(data_path, this_file), 'r', encoding='utf-8').read()
text[:500]
```

```python
# Extraire les mots clés de ce texte
keywords = kw_extractor.extract_keywords(text)
keywords
```

```python
# Ne garder que les bigrammes
kept = []
for kw, score in keywords:
    words = kw.split()
    if len(words) == 2:
        kept.append(kw)
kept
```

## Faire la même opération sur tous les documents

```python
# Extraire les mots-clés des 15 fichiers sélectionnés en 1960

for f in fichiers_analyse:
    text = open(os.path.join(data_path, f), 'r', encoding="utf-8").read()
    keywords = kw_extractor.extract_keywords(text)
    kept = []
    for kw, score in keywords:
        words = kw.split()
        if len(words) == 2:
            kept.append(kw)
    print(f"{f} mentions these keywords: {', '.join(kept)}...")
```
