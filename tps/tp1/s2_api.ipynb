## Recherche de chanteuses d'opéra belges nées entre 1830 et 1870

```python
# Femmes chanteuses d'opéra belges nées entre 1830 et 1870 (je suis fan d'opéra)

requete = """
SELECT ?personne ?personneLabel ?naissance WHERE {
  ?personne wdt:P21 wd:Q6581072 .  # Sexe : femme
  ?personne wdt:P106 wd:Q2865819 . # Profession : chanteuse d'opéra
  ?personne wdt:P27 wd:Q31 .        # Nationalité : belge
  ?personne wdt:P569 ?naissance .   # Date de naissance
  FILTER(YEAR(?naissance) >= 1830 && YEAR(?naissance) <= 1870)
  SERVICE wikibase:label { bd:serviceParam wikibase:language "fr,en". }
}
LIMIT 20
"""

# Interrogation de Wikidata
reponse = requests.get(
    "https://query.wikidata.org/sparql",
    params={"query": requete, "format": "json"},
    headers={"User-Agent": "TP1-CAMille/1.0 (https://github.com/Sheurtebise/tac)"}
)

# Affichage des résultats
for personne in reponse.json()["results"]["bindings"]:
    print(personne["personneLabel"]["value"], personne["naissance"]["value"][:4])
```
