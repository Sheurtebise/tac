# TP1 : Téléchargement automatique des PDF du corpus CAMille



Dans ce notebook, nous utilisons la technique du web scraping pour récupérer automatiquement les 51 fichiers PDF du corpus CAMille mis à disposition sur le site du cours.


## Imports

```python
import os
import time
import requests
from bs4 import BeautifulSoup
```

```python
# Adresse du site CAMille
adresse = "https://max.de.wilde.web.ulb.be/camille/"

# Récupération de la page
reponse = requests.get(adresse)
page = BeautifulSoup(reponse.text, "html.parser")

# Recherche des fichiers PDF
liens_pdf = []

for lien in page.find_all("a"):
    nom = lien.get("href")
    if nom and nom.endswith(".pdf"):
        liens_pdf.append(nom)

# Nombre de fichiers trouvés
print("Nombre de PDF trouvés :", len(liens_pdf))
```

```python
print(liens_pdf[0])
```

```python
# Création du dossier pour les PDF
dossier = os.path.expanduser("~/Documents/CAMille_PDF_TP1")
os.makedirs(dossier, exist_ok=True)

# Téléchargement du premier PDF
nom = liens_pdf[0]
pdf = requests.get(adresse + nom)

# Enregistrement du PDF
with open(dossier + "/" + nom, "wb") as fichier:
    fichier.write(pdf.content)

print("PDF téléchargé :", nom)
```

```python
# Création du dossier
dossier = os.path.expanduser("~/Documents/CAMille_PDF_TP1")
os.makedirs(dossier, exist_ok=True)

# Téléchargement des 51 PDF
for nom in liens_pdf:
    pdf = requests.get(adresse + nom)
    pdf.raise_for_status()

    with open(dossier + "/" + nom, "wb") as fichier:
        fichier.write(pdf.content)

    time.sleep(0.5)

# Vérification
print("Nombre de PDF téléchargés :", len(os.listdir(dossier)))
```
